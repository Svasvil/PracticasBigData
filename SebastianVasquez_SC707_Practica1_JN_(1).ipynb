2. Resuelva el ejercicio de la presentación mediante un dataFrame. Cree un dataFrame con una fila y 5 columnas con los valores en segundos. Use apply() para generar una segunda fila con los valores transformados en horas.

```python
import pandas as pd
import numpy as np
Segundos = np.array([3600,7200,1800,4500,10000]).reshape(1, -1) # el reshape para lo que nos funciona es precisamente para decirle a numpy que si o si el array es de 1x5

Horas = pd.DataFrame(Segundos, index=['Segundos']) #esto lo q hace es que le pone "segundos " de nombre a la linea con los datos
def seg_horas (segundos):
    return segundos / 3600
horas2 = Horas.apply(seg_horas)
print(horas2)

Horas.loc['Horas'] = Horas.iloc[0] # esto lo que hace es vamos a tomar los datos de la posicion 0 del array
#de tabla pasada que hicimos en este caso 'segundos'
print(Horas)
```
